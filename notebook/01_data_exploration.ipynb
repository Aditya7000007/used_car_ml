```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_columns = [
    "car_name",
    "brand",
    "model",
    "seller_type",
    "fuel_type",
    "transmission_type"
]

numerical_columns = [
    "vehicle_age",
    "km_driven",
    "mileage",
    "engine",
    "max_power",
    "seats"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numerical_columns),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_columns)
    ]
)

print("Preprocessor created successfully!")
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# 1. Load cleaned dataset
data_path = "../data/cardekho_dataset.csv" if os.path.exists("../data/cardekho_dataset.csv") else "data/cardekho_dataset.csv"
df = pd.read_csv(data_path)

if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])
df = df.drop_duplicates()

# Handle seats == 0
df.loc[df["seats"] == 0, "seats"] = np.nan
df["seats"] = df["seats"].fillna(df["seats"].median())

# 2. Define Buyer Risk target (0 = Lower Risk, 1 = Higher Risk)
df["buyer_risk"] = (
    (df["vehicle_age"] >= 8) &
    (df["km_driven"] >= 70000)
).astype(int)

# 3. Features & Preprocessor
categorical_columns = [
    "car_name", "brand", "model", 
    "seller_type", "fuel_type", "transmission_type"
]
numerical_columns = [
    "vehicle_age", "km_driven", "mileage", 
    "engine", "max_power", "seats"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numerical_columns),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_columns)
    ]
)

# 4. Train / Test Split
X_risk = df.drop(["buyer_risk", "selling_price"], axis=1)
y_risk = df["buyer_risk"]

X_risk_train, X_risk_test, y_risk_train, y_risk_test = train_test_split(
    X_risk,
    y_risk,
    test_size=0.2,
    random_state=42,
    stratify=y_risk
)

# 5. Train Logistic Regression Model
risk_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LogisticRegression(max_iter=1000, random_state=42))
    ]
)

risk_model.fit(X_risk_train, y_risk_train)

# 6. Predict & Evaluate
y_risk_pred = risk_model.predict(X_risk_test)

accuracy = accuracy_score(y_risk_test, y_risk_pred)
precision = precision_score(y_risk_test, y_risk_pred)
recall = recall_score(y_risk_test, y_risk_pred)
f1 = f1_score(y_risk_test, y_risk_pred)

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}\n")

print("Classification Report:")
print(classification_report(y_risk_test, y_risk_pred, target_names=["Lower Risk (0)", "Higher Risk (1)"]))

# 7. Confusion Matrix Plot
cm = confusion_matrix(y_risk_test, y_risk_pred)

fig, ax = plt.subplots(figsize=(6, 5))
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Lower Risk (0)", "Higher Risk (1)"]
)
disp.plot(cmap="Blues", values_format="d", ax=ax)
plt.title("Confusion Matrix - Logistic Regression (Buyer Risk)")
plt.grid(False)
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 1. Select the core vehicle specification features
cluster_features = [
    "vehicle_age",
    "km_driven",
    "mileage",
    "engine",
    "max_power",
    "seats"
]

X_cluster = df[cluster_features].copy()

# 2. Scale features using StandardScaler
cluster_scaler = StandardScaler()
X_cluster_scaled = cluster_scaler.fit_transform(X_cluster)

# 3. Compute Inertia (WCSS) across K = 2 to 8
inertias = []
k_range = range(2, 9)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_cluster_scaled)
    inertias.append(kmeans.inertia_)

# 4. Plot the Elbow Curve
plt.figure(figsize=(8, 4))
plt.plot(k_range, inertias, marker='o', color='#1f77b4', linewidth=2)
plt.title("Elbow Method for Optimal K (Used Car Segments)")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia (Within-Cluster Sum of Squares)")
plt.xticks(k_range)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
from sklearn.neighbors import NearestNeighbors

# 1. Fit KMeans with optimal K = 4
kmeans_model = KMeans(n_clusters=4, random_state=42, n_init=10)
df["cluster"] = kmeans_model.fit_predict(X_cluster_scaled)

# 2. Inspect cluster profiles (averages per segment)
cluster_profile = df.groupby("cluster")[cluster_features + ["selling_price"]].mean()
cluster_profile["car_count"] = df["cluster"].value_counts()
print("=== 4 Market Segment Profiles ===")
display(cluster_profile.round(1))

# 3. Assign human-readable segment names
segment_map = {
    0: "Modern Daily Commuter (Hatchback/Compact)",
    1: "Luxury & High-Performance Vehicle",
    2: "7-Seater Family SUV / MPV",
    3: "Budget & Entry-Level Used Car"
}
df["segment_name"] = df["cluster"].map(segment_map)

# 4. Build the NearestNeighbors recommender on scaled space
nn_recommender = NearestNeighbors(n_neighbors=5, metric="euclidean")
nn_recommender.fit(X_cluster_scaled)

def recommend_similar_cars(car_specs_dict, top_n=5):
    """
    Given vehicle specifications, returns top_n most similar cars in the dataset.
    """
    # Format input into dataframe matching feature columns
    input_df = pd.DataFrame([car_specs_dict])[cluster_features]
    
    # Scale input using the fitted cluster_scaler
    input_scaled = cluster_scaler.transform(input_df)
    
    # Find nearest neighbors
    distances, indices = nn_recommender.kneighbors(input_scaled, n_neighbors=top_n)
    
    # Retrieve matching records
    results = df.iloc[indices[0]][[
        "car_name", "brand", "selling_price", 
        "vehicle_age", "km_driven", "mileage", "segment_name"
    ]].copy()
    
    results["similarity_distance"] = distances[0].round(3)
    return results

# 5. Test recommendation for a sample vehicle:
# (5-year-old compact car, 45,000 km, 18 kmpl, 1197 cc, 83 bhp, 5 seats)
sample_car = {
    "vehicle_age": 5,
    "km_driven": 45000,
    "mileage": 18.0,
    "engine": 1197,
    "max_power": 83.0,
    "seats": 5
}

print("\n=== Recommendations for Sample Car ===")
sample_recommendations = recommend_similar_cars(sample_car, top_n=5)
display(sample_recommendations)
```
